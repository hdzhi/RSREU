
# Парадокс Бертрана




## Аналитическое доказательство

Пусть радиус окружности равен R. Хорда длиннее стороны вписанного равностороннего треугольника со стороной $s=\sqrt{3}R$ тогда и только тогда, когда расстояние $d$ от центра до хорды меньше $R/2$. Это следует из соотношений для длины хорды:
$$L=2\sqrt{R^2-d^2}>\sqrt{3}R \iff d<\frac{R}{2}.$$



### Метод 1: случайные концы
Метод «случайных концов»: наудачу выберем две точки на окружности и проведём через них хорду. Чтобы посчитать искомую вероятность, представим, что треугольник повёрнут так, что одна из его вершин совпадает с концом хорды. Заметим, что если другой конец хорды лежит на дуге между двумя другими вершинами треугольника, то длина хорды больше стороны треугольника. Длина рассмотренной дуги равна трети длины окружности, следуя классическому определению, искомая вероятность равна 1/3\
\
При выборе двух точек на окружности вероятность того, что центральный угол между ними меньше $\frac{2\pi}{3}$ равна $1/3$, что даёт ответ $1/3$.



### Метод 2: случайный радиус
Метод «случайного радиуса»: зафиксируем радиус окружности, наудачу выберем точку на радиусе. Построим хорду, перпендикулярную зафиксированному радиусу, проходящую через выбранную точку. Для нахождения искомой вероятности представим, что треугольник повёрнут так, что одна из его сторон перпендикулярна зафиксированному радиусу. Хорда длиннее стороны треугольника, если её центр ближе к центру, чем точка пересечения треугольника с зафиксированным радиусом.\
\
Если выбирать точку на радиусе равномерно по расстоянию от центра, то $d$ равномерно в $[0,R]$ и $P(d<R/2)=1/2$.



### Метод 3: случайный центр (середина хорды)
Метод «случайного центра»: выберем наудачу произвольную точку внутри круга и построим хорду с центром в выбранной точке. Хорда длиннее стороны равностороннего треугольника, если выбранная точка находится внутри круга, вписанного в треугольник. \
\
Если середина хорды выбирается равномерно по площади круга радиуса $R$, то условие $d<R/2$ эквивалентно попаданию в круг радиуса $R/2$, и вероятность равна $(R/2)^2/R^2=1/4$.


```python
import numpy as np
import pandas as pd
import math

R = 1.0

# Метод 1 (случайные концы): P = 1/3
# Метод 2 (случайный радиус): P = 1/2
# Метод 3 (случайная середина): P = 1/4

analytic = pd.DataFrame({
    "Метод": ["Случайные концы", "Случайный радиус", "Случайная середина"],
    "Аналитическая вероятность": [1/3, 1/2, 1/4]
})

print("Аналитические результаты")
print(analytic)


def sim_endpoints(N):
    cnt = 0
    for _ in range(N):
        t1 = np.random.uniform(0, 2*math.pi)
        t2 = np.random.uniform(0, 2*math.pi)
        ang = abs(t1 - t2)
        ang = min(ang, 2*math.pi - ang)
        L = 2*R*math.sin(ang/2)
        if L > math.sqrt(3)*R:
            cnt += 1
    return cnt / N


def sim_radius(N):
    cnt = 0
    for _ in range(N):
        d = np.random.uniform(0, R)
        if d < R/2:
            cnt += 1
    return cnt / N


def sim_midpoint(N):
    cnt = 0
    for _ in range(N):
        u = np.random.uniform(0, 1)
        d = R * math.sqrt(u)
        if d < R/2:
            cnt += 1
    return cnt / N

Nsim = 200000

simulated = pd.DataFrame({
    "Метод": ["Случайные концы", "Случайный радиус", "Случайная середина"],
    "Численная вероятность": [
        sim_endpoints(Nsim),
        sim_radius(Nsim),
        sim_midpoint(Nsim)
    ]
})

print("\nЧисленные результаты")
print(simulated)

comparison = analytic.merge(simulated, on="Метод")
print("\nСравнение аналитики и симуляции")
print(comparison)
```

## Вывод

Аналитически и численно получаем ответы: 1/3, 1/2 и 1/4 — парадокс возникает из-за неоднозначности термина «случайная хорда».
