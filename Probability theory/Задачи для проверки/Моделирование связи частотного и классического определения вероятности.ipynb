## Моделирование связи частотного и классического определения вероятности

#### В коробке находится 13 красных и 8 синих шариков. Сколько существует способов вынуть 11 красных и 6 синих шариков?

##### Пользуясь классическим определением, получим:
Рассмотрим событие А, при котором из коробки вытащили 11 красных и 6 синих шариков.

$P(A)=\frac{C_{13}^{11}\cdot C_{6}^{8}}{C_{21}^{17}}=0.3649$ 

Оценим вероятность вынуть заданное количество шариков, используя от 10 до 1000 испытаний

```python
from random import sample
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
def random_comb():
    urna = ['Red'] * 13 + ['Blue'] * 8  
    return sample(urna, 17)  

N_values = np.array(range(10, 10001, 50))
P_A_values = []

for N in N_values:
    N_A = 0
    for _ in range(N):
        comb = random_comb()
        if comb.count('Red') == 11:  
            N_A += 1
    P_A_values.append(N_A / N)

print(round(P_A_values[-1], 4))

P = [0.3649] * len(N_values) 

plt.figure(figsize=(8, 4))
plt.plot(N_values, P_A_values, label='Экспериментальная частота', alpha=0.7)
plt.axhline(y=104/285, color='red', linestyle='--', label='Теоретическая вероятность')
plt.xlabel('Количество экспериментов (N)')
plt.ylabel('Вероятность P(A)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
